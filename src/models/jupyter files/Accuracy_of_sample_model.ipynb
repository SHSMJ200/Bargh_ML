```python
from main import *

l_min = 4
max_diff = 3
c_thresh = 0.9

df = add_features_and_filter(l_min, max_diff, c_thresh, read_from_integrated=False, write_on_csv=False)
ds = Data_selector(df)
df_modified = ds.select_peaks(goodness=3)

train_errors = []
test_errors = []

power_plants = df_modified[['name', 'code']].drop_duplicates()
for _, row in power_plants.iterrows():
    df_modified2 = Data_selector(df_modified).filter_name_code(row['name'], row['code'])
    
    feature_selector = Feature_selector(df_modified2, target="generation")
    features_to_be_select = ["name", "code", "temperature"]
    feature_selector.select(features_to_select=features_to_be_select)
    X, y = feature_selector.get_X_and_y()
    
    model = Linear()
    model.scale_and_split_data(X, y)
    model.fit()
    
    rmse_error_train, rmse_error_test = model.compute_rmse_error()
    train_errors.append(rmse_error_train)
    test_errors.append(rmse_error_test)
    
import matplotlib.pyplot as plt

plt.hist(train_errors, bins=30, alpha=0.5, label='Train Errors')
plt.hist(test_errors, bins=30, alpha=0.5, label='Test Errors')
plt.legend()
plt.show()
```
