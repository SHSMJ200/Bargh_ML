```python
def set_minus(df1,df2,column=None):
    if column == None:
        column = df1.columns.to_list()
    merged = df1.merge(df2, on=column, how='left', indicator=True)
    result = merged[merged['_merge'] == 'left_only'].drop('_merge', axis=1)
    return result
def equals(df1,df2):
    return set_minus(df1,df2).empty and set_minus(df2,df1).empty
```

```python
def delete_C(df):
    df1 = df.copy()
    df1["code_e"] = df1["code"].copy()
    df1["code"] = df1["code"].apply(lambda x: x[1:] if x.startswith("C") else x)
    return df1
```
