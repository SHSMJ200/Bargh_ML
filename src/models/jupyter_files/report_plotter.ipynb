```python
import os
import sys

current_dir = os.getcwd()
project_root = current_dir[:current_dir.find("src") - 1]
sys.path.insert(0, project_root)
```

```python
import plotly.graph_objects as go
from src.models.filter_data.filter_data import *
from src.models.filter_data.feature_adder import *
```

```python
csv_read_path = os.path.join(project_root, "data", "processed", "semi_processed.csv")
df = pd.read_csv(csv_read_path, encoding='utf-8')
```

```python
def draw_gen_temp_plot(df_c, x_col, y_col, name, code, color_map, save=False, project_root=None, title=""):
    fig = go.Figure()
    df_show = df_c[(df_c["name"] == name) & (df_c["code"] == code)]
    
    for peak_value, color in color_map.items():
        df_subset = df_show[df_show["is_good_peak"] == peak_value]
        fig.add_trace(go.Scatter(
            x=df_subset[x_col],
            y=df_subset[y_col],
            mode="markers",
            marker=dict(size=4, color=color),
            name=f"is_good_peak = {peak_value}",
            hovertext=df_subset["datetime"]
        ))

    fig.update_traces(marker=dict(size=4, sizemode="diameter", sizeref=1, opacity=0.7))

    if title == "turbo" and name == "قم" and code == "G11":
        a = -0.689
        b = 119.66736878408216
        x_line = np.linspace(df_show[x_col].min(), df_show[x_col].max(), 100)
        df_line = pd.DataFrame(x_line, columns=[x_col])
        y_line = a * x_line + b
        fig.add_trace(go.Scatter(
            x=df_line[x_col],
            y=y_line,
            mode="lines",
            line=dict(dash="solid", width=2),
            visible="legendonly"
        ))

    if save and project_root:
        save_path = f"{project_root}/src/visualization/unit_figs/report/{name}-{code}_{title}.html"
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        fig.write_html(save_path)
    else:
        fig.show()
```

```python
color_map = {
    1: "purple",
    2: "orange",
    3: "pink",
    4: "red",
    5: "black",
    6: "red"
}
name = "پرند"
code = "G11"
draw_gen_temp_plot(df, x_col="datetime", y_col="generation", name=name, code=code, color_map=color_map, save=True,
                   project_root=project_root, title="date")
```
